# extractrs vs exactextract — Validation

Validates that extractrs produces **bit-identical** results to exactextract on real climate data.

- **Raster**: Daymet V4R1 daily tmin (local NetCDF)
- **Basins**: MERIT Hydro Pfafstetter level-7 basins (local shapefile)

```python
import os
import tempfile
import time
import warnings

import exactextract
import extractrs as extrs
import geopandas as gpd
import numpy as np
import pandas as pd
import rasterio
import rioxarray  # noqa: F401 — registers .rio accessor
import xarray as xr
from rasterio.transform import from_origin
from rasterio.windows import from_bounds
```

## Configuration

```python
NC_PATH = "/mnt/ssd1/data/daymet/daymet_v4_daily_na_tmin_2024.nc"
SHP_PATH = "/mnt/ssd1/data/merit/cat_pfaf_7_MERIT_Hydro_v07_Basins_v01_bugfix1.shp"
BBOX = (-84.5, 34.5, -84.0, 35.0)  # WGS84: west, south, east, north
VAR = "tmin"
TIME_IDX = 182  # July 1 (0-indexed)
ID_COL = "COMID"
N_BENCH = 10

BAD_COMIDS = {74009117}  # known to segfault exactextract

DAYMET_CRS = (
    "+proj=lcc +lat_0=42.5 +lon_0=-100 +lat_1=25 +lat_2=60 "
    "+x_0=0 +y_0=0 +datum=WGS84 +units=m"
)
```

## Load basins

```python
basins = gpd.read_file(SHP_PATH, bbox=BBOX)
basins = basins.set_crs("EPSG:4326", allow_override=True)
basins = basins[~basins[ID_COL].isin(BAD_COMIDS)]
print(f"{len(basins)} basins loaded")
basins.head()
```

## Load Daymet

```python
ds = xr.open_dataset(NC_PATH)
ds = ds[[VAR]].isel(time=TIME_IDX)
ds = ds.rio.write_crs(DAYMET_CRS)
print(f"Dataset: {dict(ds.sizes)}")
```

## Run exactextract

Create a grid-aligned GeoTIFF subset for exactextract, then compute zonal means.

```python
# Reproject basins to Daymet LCC for exactextract
basins_proj = basins.to_crs(DAYMET_CRS)

# Build a grid-aligned GeoTIFF subset from the xarray data
da = ds[VAR]
x = da["x"].values.astype(np.float64)
y = da["y"].values.astype(np.float64)
dx, dy = abs(float(x[1] - x[0])), abs(float(y[1] - y[0]))

full_transform = from_origin(float(x.min()) - dx / 2, float(y.max()) + dy / 2, dx, dy)

# Window around basins, snapped to pixel boundaries
tb = basins_proj.total_bounds
pad = 2000
window = from_bounds(tb[0] - pad, tb[1] - pad, tb[2] + pad, tb[3] + pad, full_transform)
window = window.round_offsets().round_lengths(op="ceil")

row_off, col_off = int(window.row_off), int(window.col_off)
height, width = int(window.height), int(window.width)

data_subset = da.values[row_off:row_off + height, col_off:col_off + width]
win_transform = rasterio.windows.transform(window, full_transform)

tif_path = os.path.join(tempfile.mkdtemp(), "daymet_subset.tif")
with rasterio.open(
    tif_path, "w", driver="GTiff",
    height=height, width=width, count=1,
    dtype=data_subset.dtype, crs=DAYMET_CRS,
    transform=win_transform, nodata=-9999.0,
) as dst:
    dst.write(data_subset, 1)

rast_sub = rasterio.open(tif_path)
print(f"GeoTIFF: {rast_sub.width}x{rast_sub.height}")
```

```python
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    ee_result = exactextract.exact_extract(
        rast_sub, basins_proj,
        ops=["mean"],
        include_cols=[ID_COL],
        output="pandas",
    )

rast_sub.close()
ee_result.head()
```

## Run extractrs

```python
rs_result = ds.extrs.zonal_stats(basins, stat="mean", id_col=ID_COL)
rs_result
```

## Compare results

```python
ee_df = ee_result.rename(columns={"mean": "exactextract"})
rs_df = pd.DataFrame({
    ID_COL: rs_result[ID_COL].values,
    "extractrs": rs_result[VAR].values,
})

merged = ee_df[[ID_COL, "exactextract"]].merge(rs_df, on=ID_COL)
merged = merged.dropna(subset=["exactextract", "extractrs"])
merged["abs_diff"] = (merged["exactextract"] - merged["extractrs"]).abs()

corr = merged["exactextract"].corr(merged["extractrs"])
print(f"Basins:    {len(merged)}")
print(f"Pearson R: {corr:.10f}")
print(f"Max diff:  {merged['abs_diff'].max():.8f} C")
print(f"Mean diff: {merged['abs_diff'].mean():.8f} C")
merged
```

## Benchmark

Compare per-run performance after the one-time cache build.

```python
# exactextract timing
rast_sub = rasterio.open(tif_path)
ee_times = []
for _ in range(N_BENCH):
    t0 = time.time()
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        exactextract.exact_extract(rast_sub, basins_proj, ops=["mean"],
                                   include_cols=[ID_COL], output="pandas")
    ee_times.append(time.time() - t0)
rast_sub.close()

# extractrs cache build + apply_stat timing
data_2d = ds[VAR].values.astype(np.float64)
nodata = float(ds[VAR].encoding.get("_FillValue", -9999.0))

t0 = time.time()
cache = extrs.build_cache(
    [g.wkb for g in basins_proj.geometry],
    basins_proj[ID_COL].astype(np.int64).tolist(),
    float(x.min()) - dx / 2, float(y.min()) - dy / 2,
    float(x.max()) + dx / 2, float(y.max()) + dy / 2, dx, dy,
)
cache_time = time.time() - t0

rs_times = []
for _ in range(N_BENCH):
    t0 = time.time()
    extrs.apply_stat(cache, data_2d, nodata, "mean")
    rs_times.append(time.time() - t0)

ee_ms = np.median(ee_times) * 1000
rs_ms = np.median(rs_times) * 1000
print(f"exactextract:  {ee_ms:.1f} ms/run")
print(f"extractrs:")
print(f"  cache build: {cache_time*1000:.1f} ms (one-time)")
print(f"  apply_stat:  {rs_ms:.3f} ms/run")
if rs_ms > 0:
    print(f"  speedup:     {ee_ms/rs_ms:.0f}x")
```
